Imports

```python
from sklearn.ensemble import RandomForestRegressor
import numpy as np
import pandas as pd 
```

Process data

```python
train_data = pd.read_csv("all_seasons.csv")
train_data.head()
```

```python
# Use points scored as the target for regression
y = train_data['pts']
```

```python
# Choose which features to focus on
features = ["Age", "G", "MP", "PTS", "TRB", "AST", "STL", "BLK", "FG%", "3P%", "FT%"]

# Create training matrix
X = pd.get_dummies(train_data[features])
```

Create Ranadom Forest

```python
# Use a RandomForest regressor to predict numeric points
model = RandomForestRegressor(n_estimators=100, max_depth=5, random_state=1)
# Train the model
model.fit(X, y)
```

Test data

```python
test_data = pd.read_csv("mvp.csv")
test_data.head()

X_test = pd.get_dummies(test_data[features])

# Predict with test data
predictions = model.predict(X_test)
```

```python
# Save predictions with a sensible column name for this dataset
output = pd.DataFrame({'player_name': test_data.player_name, 'predicted_pts': predictions})
output.to_csv('submission.csv', index=False)
print("Predictions saved to submission.csv")
```
